```python
import pandas as pd

df = pd.read_csv("supermarket_data.csv")

# Calculate descriptive statistics for numeric columns
mean_price = df['itemprice'].mean()
median_price = df['itemprice'].median()
variance_price = df['itemprice'].var()
std_price = df['itemprice'].std()

print("Mean Price:", mean_price)
print("Median Price:", median_price)
print("Variance of Price:", variance_price)
print("Std Deviation of Price:", std_price)
```

```python
import numpy as np
from scipy.stats import *

# Normal Distribution - Using customer salary
salary_mean = df['customersalary'].mean()
salary_std = df['customersalary'].std()
print("Normal Dist (Salary) ~ Mean:", salary_mean, "Std:", salary_std)

# Binomial Distribution - Gender (Male/Female)
female_count = (df['gender'] == 'Female').sum()
total_count = len(df)
prob_female = female_count / total_count
print("Probability of Female Buyer (Binomial):", prob_female)

# Uniform Distribution - Generate 10 random offers between 5% to 20% discount
uniform_offers = uniform.rvs(loc=5, scale=15, size=10)
print("Uniform Random Discounts:", uniform_offers)

```

```python
import numpy as np
from scipy import stats

# Take a sample of 50 total purchase values
sample = df['total'].sample(50, random_state=42)

sample_mean = sample.mean()
sample_std = sample.std()
n = len(sample)
z = 1.96  # 95% confidence

# Margin of Error & Confidence Interval
margin = z * (sample_std / np.sqrt(n))
ci_lower, ci_upper = sample_mean - margin, sample_mean + margin

print("Sample Mean:", sample_mean)
print("95% Confidence Interval:", (ci_lower, ci_upper))
```

```python
from scipy.stats import ttest_ind

male_total = df[df['gender']=='Male']['total']
female_total = df[df['gender']=='Female']['total']

t_stat, p_value = ttest_ind(male_total, female_total)
print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

table = pd.crosstab(df['gender'], df['product'])
chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square Value:", chi2)
print("P-Value:", p)
print("Expected Table:\n", expected)
```
