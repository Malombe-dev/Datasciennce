```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

from scipy.stats import norm
z = norm.ppf(0.95)
print(z)   # 1.645
# z-table
# MOE = z * (sigma/sqrt(n)) that is confidence level /sqrt(n)
# Yes! There is a Z-table (Standard Normal Table) that provides values like 1.645, 1.96, 2.576 based on the area under the normal distribution curve.
```

```python
# Simulate population (heights in cm)
population = np.random.normal(170, 10, 10000)


# Take a random sample of 50
sample = np.random.choice(population, 50, replace=False)
print("Population Mean:", np.mean(population))
print("Sample Mean:", np.mean(sample))
```

```python
sample_mean = np.mean(sample)
print("Sample Mean:", sample_mean)
# 👉 Delta Degrees of Freedom
# “So it simply means we are using sample data, we adjust the formula.
sample_std = np.std(sample, ddof=1)
n = len(sample)
z = 1.96 # 95% confidence
margin = z * (sample_std / np.sqrt(n))
print("95% CI:", (sample_mean - margin, sample_mean + margin))
```

```python
ci = stats.t.interval(0.95, df=n-1, loc=sample_mean, scale=sample_std/np.sqrt(n))
print("SciPy CI:", ci)
```

```python
plt.axvline(sample_mean, label='Sample Mean')
plt.axvspan(sample_mean-margin, sample_mean+margin, alpha=0.3, label='95% CI')
plt.title('Confidence Interval Visualization')
plt.legend()
plt.show()
# When you see a vertical axis going up to 1 in a confidence interval or distribution plot, it usually means the plot is showing a probability density function (PDF) — not actual data counts.
```
