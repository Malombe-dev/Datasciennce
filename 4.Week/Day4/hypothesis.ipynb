```python
import numpy as np
from scipy import stats

# Sample data: Weight before and after diet
group1 = np.array([72, 75, 78, 71, 74])  # Before
group2 = np.array([70, 73, 74, 69, 72])  # After

# Perform Independent t-test
t_stat, p_value = stats.ttest_ind(group1, group2)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python
import numpy as np
from scipy.stats import chi2_contingency

# Gender vs Product Preference
# Rows: Male, Female | Columns: Shoes, Clothes, Electronics
data = np.array([[40, 25, 35],
                 [30, 50, 20]])

chi2, p, dof, expected = chi2_contingency(data)

print("Chi-Square Value:", chi2)
print("P-Value:", p)
print("Degrees of Freedom:", dof)
print("Expected Table:\n", expected)

alpha = 0.05
if p < alpha:
    print("\nConclusion: Reject Null Hypothesis – Gender affects product preference.")
else:
    print("\nConclusion: Fail to Reject Null – No significant difference.")

```

```python
# Product preference data
# Rows: Male, Female
# Columns: Shoes, Clothes, Electronics
import pandas as pd

data = np.array([
    [40, 25, 35],   # Male
    [30, 50, 20]    # Female
])

df = pd.DataFrame(data, 
                  index=["Male", "Female"], 
                  columns=["Shoes", "Clothes", "Electronics"])

print(df)
```

```python
# Visualization
import matplotlib.pyplot as plt
categories = ["Shoes", "Clothes", "Electronics"]
male = [40, 25, 35]
female = [30, 50, 20]

x = np.arange(len(categories))
width = 0.35

plt.bar(x - width/2, male, width, label="Male")
plt.bar(x + width/2, female, width, label="Female")
plt.xticks(x, categories)
plt.title("Product Preference by Gender")
plt.xlabel("Product Category")
plt.ylabel("Number of Customers")
plt.legend()
plt.show()
```
