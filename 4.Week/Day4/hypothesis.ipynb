```python
import numpy as np
from scipy import stats

# Sample data: Weight before and after diet
group1 = np.array([72, 75, 78, 71, 74])  # Before
group2 = np.array([70, 73, 74, 69, 72])  # After

# Perform Independent t-test
# t_stat → T-statistic (difference between the means in units of variation)
# p_value → Probability of difference occurring by chance
# ttest_ind = Independent T-test
t_stat, p_value = stats.ttest_ind(group1, group2)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)

# ✅ Decision Based on P-Value
# P-Value = 0.1568
# Common significance level (α) = 0.05
# 📌 Rule:
# If P < 0.05 → Significant difference
# If P > 0.05 → No significant difference
# 👉 Here, 0.1568 > 0.05
```

```python
import numpy as np
from scipy.stats import chi2_contingency

# Gender vs Product Preference
# Rows: Male, Female | Columns: Shoes, Clothes, Electronics
data = np.array([[40, 25, 35],
                 [30, 50, 20]])

# chi2 → Chi-Square test statistic
# p → P-value (probability of results happening by chance)
# dof → Degrees of freedom
# expected → Expected frequency table if there was no relationship (null hypothesis true)             

chi2, p, dof, expected = chi2_contingency(data)

print("Chi-Square Value:", chi2)
print("P-Value:", p)
print("Degrees of Freedom:", dof)
print("Expected Table:\n", expected) #asuming no gender difference 

alpha = 0.05
if p < alpha:
    print("\nConclusion: Reject Null Hypothesis – Gender affects product preference.")
else:
    print("\nConclusion: Fail to Reject Null – No significant difference.")

```

```python
# Product preference data
# Rows: Male, Female
# Columns: Shoes, Clothes, Electronics
import pandas as pd

import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Gender": ["Male", "Male", "Male", "Female", "Female", "Female"],
    "Product": ["Shoes", "Clothes", "Electronics", "Shoes", "Clothes", "Electronics"],
    "Count": [40, 25, 35, 30, 50, 20]
})


print(df)
```

```python
grouped = df.groupby(['Gender', 'Product'])['Count'].sum().unstack()

grouped.plot(kind='bar')
plt.title("Product Preference by Gender (Using GroupBy)")
plt.ylabel("Number of Customers")
plt.show()
```
